#How did italian banks react to ECB rate decision?
Event study on 5 Italian listed banks, July 2022-2025.


```python
!pip install yfinance -q
import yfinance as yf

tickers = ["ISP.MI", "UCG.MI", "BAMI.MI", "BPE.MI", "BMPS.MI", "FTSEMIB.MI"]
prices = yf.download(tickers, start="2022-07-01", auto_adjust=True)["Close"]
prices.tail()
```

```python
# Data quality check: date range, number of trading days, missing values
print(f"Period: {prices.index.min().date()} to {prices.index.max().date()}")
print(f"Trading days: {len(prices)}")
print("\nMissing values per column:")
print(prices.isna().sum())
```

### Data quality check
No missing values: daily prices are complete for all 5 banks and the FTSE MIB index over the whole period.

## ECB monetary policy decisions (July 2022 – December 2025)
All 28 monetary policy meetings in the period, including those where rates were left unchanged.
The reference rate is the **deposit facility rate (DFR)**, the ECB's main policy rate since 2022.
Decisions are announced at 14:15 CET, during trading hours, so the event day (day 0) is the announcement day.

Source: ECB press releases, *Monetary policy decisions* (ecb.europa.eu).

```python
import io
import pandas as pd

ecb_csv = """date,decision,change_bp,deposit_rate_after
2022-07-21,hike,50,0.00
2022-09-08,hike,75,0.75
2022-10-27,hike,75,1.50
2022-12-15,hike,50,2.00
2023-02-02,hike,50,2.50
2023-03-16,hike,50,3.00
2023-05-04,hike,25,3.25
2023-06-15,hike,25,3.50
2023-07-27,hike,25,3.75
2023-09-14,hike,25,4.00
2023-10-26,hold,0,4.00
2023-12-14,hold,0,4.00
2024-01-25,hold,0,4.00
2024-03-07,hold,0,4.00
2024-04-11,hold,0,4.00
2024-06-06,cut,-25,3.75
2024-07-18,hold,0,3.75
2024-09-12,cut,-25,3.50
2024-10-17,cut,-25,3.25
2024-12-12,cut,-25,3.00
2025-01-30,cut,-25,2.75
2025-03-06,cut,-25,2.50
2025-04-17,cut,-25,2.25
2025-06-05,cut,-25,2.00
2025-07-24,hold,0,2.00
2025-09-11,hold,0,2.00
2025-10-30,hold,0,2.00
2025-12-18,hold,0,2.00
"""

ecb = pd.read_csv(io.StringIO(ecb_csv), parse_dates=["date"])
ecb.to_csv("ecb_decisions_2022_2025.csv", index=False)

print(ecb["decision"].value_counts())
ecb
```
